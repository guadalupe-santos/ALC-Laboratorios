EJERCICIO 1: Recibe dos numeros x e y, y calcula el error de aproximar x usando y en float64

```python
def error(x,y):
  y = np.float64(y)
  return np.absolute(np.float64(x-y))

error(np.float32(0.3),0.3)
```

EJERCICIO 2: Recibe dos numeros x e y, y calcula el error relativo de aproximar x usando y en float64

```python
def error_relativo(x,y):
  if x != 0:
    return (error(x,y))/ np.absolute(x)

error_relativo(0.25,0.3)
```
