```python
import numpy as np
def rota(θ):
  a = np.cos(θ)
  b = np.sin(θ)
  T = np.array([[a,-b],[b,a]])
  return T
rota(0.5*np.pi)
```

```python
def escala(s):
  T = []
  for i in range(0,len(s)):
    fila = []
    for j in range(0,len(s)):
      if j == i:
        fila.append(s[j])
      else:
        fila.append(0)
    T.append(fila)
  return np.array(T)
escala([2,3])
```

```python
def rota_y_escala(θ,s):
  T = escala(s) @ rota(θ)
  return T
rota_y_escala(0.5*np.pi,[2,3])
```

```python
def afin(θ,s,b):
  A = rota_y_escala(θ,s)
  T = np.array([[A[0,0],A[0,1],b[0]],
                [A[1,0],A[1,1],b[1]],
                 [0,0,1]])
  return T


afin(0.5*np.pi,[2,3],[1,2])
```

```python
def trans_afin(v,θ,s,b):
  T = afin(θ,s,b)
  #Corrigo dimensión la dimensión del vector
  v3 = np.append(v,1)
  x = T[0]@v3
  y = T[1]@v3
  xy = []
  xy.append(x)
  xy.append(y)
  return xy

trans_afin([1,2],0.5*np.pi,[2,3],[1,2])
```

```python
#TEST
assert(np.allclose(rota(0), np.eye(2)))
assert(np.allclose(rota(np.pi/2), np.array([[0, -1],[1, 0]])))
assert(np.allclose(rota(np.pi), np.array([[-1, 0],[0, -1]])))

# Tests para escala
assert(np.allclose(escala([2,3]), np.array([[2,0],[0,3]])))
assert(np.allclose(escala([1,1,1]), np.eye(3)))
assert(
    np.allclose(escala([0.5,0.25]), np.array([[0.5,0],[0,0.25]]))
)

# Tests para rota_y_escala
assert(
    np.allclose(rota_y_escala(0,[2,3]), np.array([[2,0],[0,3]]))
)
assert(np.allclose(
    rota_y_escala(np.pi/2,[1,1]), np.array([[0,-1],[1,0]])
))
assert(np.allclose(
    rota_y_escala(np.pi,[2,2]), np.array([[-2,0],[0,-2]])
))

# Tests para afin
assert(np.allclose(
    afin(0,[1,1],[1,2]),
    np.array([[1,0,1],
              [0,1,2],
              [0,0,1]])))
assert(np.allclose(afin(np.pi/2,[1,1],[0,0]),
    np.array([[0,-1,0],
              [1,0,0],
              [0,0,1]])))
assert(np.allclose(afin(0,[2,3],[1,1]),
    np.array([[2,0,1],
              [0,3,1],
              [0,0,1]])))

# Tests para trans_afina

assert(np.allclose(
    trans_afin(np.array([1,0]), np.pi/2,[1,1],[0,0]),
    np.array([0,1])
))
assert(np.allclose(
    trans_afin(np.array([1,1]), 0,[2,3],[0,0]),
    np.array([2,3])
))
assert(np.allclose(
    trans_afin(np.array([1,0]), np.pi/2,[3,2],[4,5]),
    np.array([4,7])
))

print("¡Todos los tests del Laboratorio 2 han pasado exitosamente!")
```
