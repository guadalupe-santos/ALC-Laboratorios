Ejercicio 1

```python
import numpy as np
#p puede ser infinito, en ese caso el módulo es el maximo
def norma(x,p):
  res = 0
  if p == 'inf':
    max = 0
    for j in range(len(x)):
      if abs(x[j]) > max:
        max = abs(x[j])
    return max
  for i in range(len(x)):
    res = res + (abs(x[i]))**p
  return res**(1/p)
norma([-1,8,7],3)
```

Ejercicio 2

```python
def normaliza(X,p):
  res = []
  for i in range(len(X)):
    normaVector = norma(X[i],p)
    v = []
    for j in range(len(X[i])):
      s = X[i][j]/normaVector
      v.append(s)
    res.append(v)
  return res
normaliza([[1,0,0],[0,2,2]],2)
```

Ejercicio 3 - Método de Monte Carlo

```python
def normaMatMC(A,q,p,Np):
  n = normaliza(np.random.random((Np, A.shape[1])),p)
  n_norm = np.array(n)
  Ax = []
  trans = []
  for i in range(n_norm.shape[0]):
    Ax = A@n_norm[i]
    trans.append(Ax)
  trans_array = np.array(trans)
  norma_max = 0
  v_max = []
  for j in range(trans_array.shape[0]):
    if norma(trans_array[j],q) > norma_max:
      norma_max = norma(trans_array[j],q)
      v_max = trans_array[j]
  return [norma_max,v_max]

```

Ejercicio 4

```python
def normaExacta(A,p= [1,'inf']):
  res = []
  max_f = 0
  for i in range(A.shape[0]):
    sum = 0
    for j in range(A.shape[1]):
      sum = sum + abs(A[i][j])
    if sum > max_f:
        max_f = sum
  max_c = 0
  for k in range(A.shape[1]):
    sum = 0
    for j in range(A.shape[0]):
      sum = sum + abs(A[j][k])
    if sum > max_c:
        max_c = sum
  res.append(max_c)
  res.append(max_f)
  if p == 1:
    return max_c
  elif p == 'inf':
    return max_f
  elif p == [1,'inf']:
    return res
```

Ejercicio 5

```python
def condMC(A,p):
  A_inv = np.linalg.inv(A)
  res = 0
  normaA = 0
  norma_inv = 0
  if p == [1,'inf']:
    normaA = normaExacta(A,p)
    norma_inv = normaExacta(A_inv,p)
    res = normaA*norma_inv
  else:
    normaA = normaMatMC(A,p,p,10000)
    norma_inv = normaMatMC(A_inv,p,p,10000)
    res = normaA[0]*norma_inv[0]
  return res

```

```python
A = np.array([[1,1],[0,1]])
A_ = np.linalg.solve(A,np.eye(A.shape[0]))
normaA = normaMatMC(A,2,2,10000)
normaA_ = normaMatMC(A_,2,2,10000)
condA = condMC(A,2,10000)
assert(np.allclose(normaA[0]*normaA_[0],condA,atol=1e-3))

A = np.array([[3,2],[4,1]])
A_ = np.linalg.solve(A,np.eye(A.shape[0]))
normaA = normaMatMC(A,2,2,10000)
normaA_ = normaMatMC(A_,2,2,10000)
condA = condMC(A,2,10000)
assert(np.allclose(normaA[0]*normaA_[0],condA,atol=1e-3))
```
